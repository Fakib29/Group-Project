```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/updated_pollution_dataset.csv')
df.head()
```

```python
df.info()
```

```python
import pandas as pd

# Assuming 'df' is your DataFrame

for column in df.columns:
    if pd.api.types.is_numeric_dtype(df[column]) and not pd.api.types.is_bool_dtype(df[column]):
        # Continuous numerical column
        mean = df[column].mean()
        median = df[column].median()
        variance = df[column].var()
        std_dev = df[column].std()
        print(f"Column: {column}")
        print(f"  Mean: {mean}")
        print(f"  Median: {median}")
        print(f"  Variance: {variance}")
        print(f"  Standard Deviation: {std_dev}")
        print("-" * 20)  # Separator
    else:
        # Categorical or discrete column
        frequency = df[column].value_counts()
        percentage = (frequency / len(df)) * 100
        print(f"Column: {column}")
        print("  Frequency:")
        print(frequency)
        print("  Percentage:")
        print(percentage)
        print("-" * 20)  # Separator
```

```python

import matplotlib.pyplot as plt

# Assuming 'df' is your DataFrame
column_name = 'Air Quality'  # Replace with the actual column name if different

plt.figure(figsize=(10, 6))  # Adjust figure size as needed
df[column_name].value_counts().plot(kind='bar')
plt.title(f"Bar Chart of {column_name}")
plt.xlabel(column_name)
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels if needed
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming 'df' is your DataFrame

# Select only numerical features for correlation analysis
numerical_features = df.select_dtypes(include=['number'])

# Calculate the correlation matrix
correlation_matrix = numerical_features.corr()

# Create the heatmap
plt.figure(figsize=(12, 10))  # Adjust figure size as needed
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
